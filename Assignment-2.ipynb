# Create a simple calculator that takes two numbers and an operator (+, -, *, /). Perform the selected operation and display an appropriate message if the operator is invalid.

```python
num_1 = float(input("enter a number")) #taking input from user 
num_2 = float(input("enter a number")) #taking input from user 
operator = input("enter operator. (+,-,/,*)")
if operator == "+":
    print(num_1 + num_2)
elif operator == "-":
    print(num_1 - num_2)
elif operator == "/":
        if num_2 !=0: # _checking the zero_division_error
            print(num_1 / num_2)
        else:
                print("error: dividion by zero")
elif operator == "*":
        print(num_1 * num_2)
else:
    print("invalid operator.")
```

# Create an electricity bill calculator using the following units:

### 0 to 100 units: Rs. 10 per unit

### 101–200 units: Rs. 15 per unit

### Above 200 units: Rs. 20 per unit

### Take the consumed units as input and calculate the bill according to the applicable rate.

```python
electric_unit = int(input("enter consumed electricity units")) #taking input from user

if electric_unit >200:
    unit_price = 20                 #unit price for above 200 units is 20
    bill = electric_unit * unit_price   #calcularing bill
elif electric_unit >100:            
    unit_price = 15                 #unit price for 101 to 200 units is 15
    bill = electric_unit * unit_price   #calcularing bill
else:
    unit_price = 10                 #unit price for under 100 units is 10
    bill = electric_unit * unit_price   #calcularing bill
    
print(f"your consumed unit for this period are {electric_unit}")
print(f"the rate per unit for {electric_unit} units is {unit_price} and your current bill is :{bill} .")
```
